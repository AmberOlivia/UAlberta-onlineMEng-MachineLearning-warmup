```python
import numpy as np

# Create special matrices
matrix_diag = np.diag([3,4,-5,2])
matrix_identity = np.identity(5)
matrix_ones = np.ones([5,5])
matrix_zeros = np.zeros([5,5])

print(matrix_diag)
print(matrix_identity)
print(matrix_ones)
print(matrix_zeros)
```

```python
import numpy as np

# Matrix operations

A = np.array([[2, 4], [5, -6]])
B = np.array([[9, -3], [3, 6]])
C = A + B      # element wise addition
D = A + B.T
E = A*B
F = A.dot(B)

print(A)
print(B)
print(C)
print(D)
print(E)
print(F)
```

```python
import numpy as np

# Access matrix elements

A = np.array([2, 4, 6, 8, 10])

print("A[0] =", A[0])     # First element     
print("A[2] =", A[2])     # Third element 
print("A[-1] =", A[-1])   # Last element     

B = np.array([[1, 4, 5, 12],
    [-5, 8, 9, 0],
    [-6, 7, 11, 19]])

#  First element of first row
print("B[0][0] =", B[0][0])  

# Third element of second row
print("B[1][2] =", B[1][2])

# Last element of last row
print("B[-1][-1] =", B[-1][-1])  

# Access rows of a Matrix
print("B[0] =", B[0]) # First Row
print("B[2] =", B[2]) # Third Row
print("B[-1] =", B[-1]) # Last Row (3rd row in this case)

# Access columns of a Matrix
print("B[:,0] =",B[:,0]) # First Column
print("B[:,3] =", B[:,3]) # Fourth Column
print("B[:,-1] =", B[:,-1]) # Last Column (4th column in this case)
```

```python
import numpy as np

# Properties of vectors and matrices

# Determinant
A = np.array([[6,1,1], [4, -2, 5], [2,8,7]]) 
print(A) 
print(np.linalg.det(A)) 
print(6*(-2*7 - 5*8) - 1*(4*7 - 5*2) + 1*(4*8 - -2*2))
```

```python
import numpy as np 

vector = np.array([2, 4, 6, 8, -10, -12, -14])

# L1 norm
print(np.linalg.norm(vector, 1))

# L2 norm
print(np.linalg.norm(vector, 2))

# L-inf norm
print(np.linalg.norm(vector, np.inf))

# rank of a matrix
A = np.array([[1,2,3], [1, 2, 3], [2,3,4]]) 
print(A)
print(np.linalg.matrix_rank(A))
```

```python
import numpy as np

# Eigenvalues and Eigenvectors

A = np.array([[0, 2], 
              [2, 3]])
W,V=np.linalg.eig(A)
print('E-value:', W)
print('E-vector', V)

B = np.array([[2, 2, 4], 
              [1, 3, 5],
              [2, 3, 4]])
W,V=np.linalg.eig(B)
print('E-value:', W)
print('E-vector', B)
```
