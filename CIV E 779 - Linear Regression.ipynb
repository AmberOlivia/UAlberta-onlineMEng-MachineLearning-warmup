```python
# Sklearn; we assume theta_0 = 0
import numpy as np
from sklearn.linear_model import LinearRegression
X = np.array([[1], [2], [3], [4]])
y = np.array([1,2,3,4])*3
# y = 3 * x_0
#print(X.shape)
#print(y.shape)
reg = LinearRegression(fit_intercept=False).fit(X, y)
reg.score(X, y)
#print(reg.coef_)
#print(reg.intercept_)
reg.predict(np.array([[10]]))
```

```python
# Gradient Descent; we assume theta_0 = 0
import numpy as np
import random

X = np.array([[1], [2], [3], [4]])
y = np.array([1,2,3,4])*3
# y = 3 * x_0

epsilon = 1e-6
theta_0 = 0
theta_1 = 0
J = 100
lr = 1e-2
criterion = 10000
m = X.shape[0]
iteration = 0

theta_1_tmp = theta_1

while criterion > epsilon and iteration < 100000:
    J_tmp = J
    theta_1_tmp = theta_1
    theta_1 = theta_1_tmp - lr * np.mean((theta_1_tmp*X[:,0] - y)*X[:,0])
    J = 1/2*np.mean((theta_1*X[:,0] - y)**2)
    criterion = np.abs((J-J_tmp)/J_tmp)
    iteration += 1
    if iteration % 10000 == 0:
        print("iteration", iteration)
        print("J:",J)
        print("theta_1:",theta_1)
        print("criterion:",criterion)

prediction = theta_1*10 
print(prediction)
```
