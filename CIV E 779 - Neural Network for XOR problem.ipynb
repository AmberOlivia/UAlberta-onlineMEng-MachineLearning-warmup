```python
from sklearn.neural_network import MLPClassifier
X_train = [[0,0],[0,1],[1,0],[1,1]]
y_train = [0, 1, 1, 0]
clf = MLPClassifier(solver='lbfgs', hidden_layer_sizes=(5,5,5), activation='logistic', alpha=1e-8, random_state=2, max_iter=1e8,tol=1e-20)
clf.fit(X_train, y_train)
clf.predict(X_train)
```
